### Are paper leaks becoming more frequent over time?

```python
print(df['date'].dtypes)
```

```python
## Date is an object right now. We have to convert it into datetime variable

df['date']= pd.to_datetime(df['date'])
```

```python
## Pandas is trying to interpret  dates as: mm-dd-yy. But date is in dd-mm-yy format

df['date'] = pd.to_datetime(df['date'], format='%d-%m-%y')
```

```python
print(df['date'].dtype)
```

```python
df['year'] = df['date'].dt.year

## dt.year is used to extract year from the date
```

```python
df[['date','year']].head()
```

```python
df['year'].value_counts().sort_index()
```

```python
df['year'].value_counts().sort_values(ascending=False)    ## sorting values in descending order
```
